```python
import matplotlib.pyplot as plt

# Datos
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

# Crear gráfico
plt.plot(x, y)

# Personalización
plt.title("Ejemplo de gráfico de líneas")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.grid(True)

# Mostrar gráfico
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# Crear un array de NumPy
array = np.array([1, 2, 3, 4, 5])
# Crear un DataFrame de Pandas
data = {'Columna1': [1, 2, 3], 'Columna2': [4, 5, 6]}
df = pd.DataFrame(data)
# Crear una gráfica simple con Matplotlib
plt.plot(array)
plt.title('Ejemplo de Gráfica')
plt.xlabel('Índice')
plt.ylabel('Valor')
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(
    x, y,
    color="red",        # Color de la línea
    linestyle="--",     # Estilo de línea
    linewidth=2,        # Grosor
    marker="o",         # Marcador
    markersize=8        # Tamaño del marcador
)

plt.title("Personalización básica")
plt.show()
```
